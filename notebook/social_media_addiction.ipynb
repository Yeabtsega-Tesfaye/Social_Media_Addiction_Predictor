```python
import pandas as pd

# Load dataset
df = pd.read_csv("data/dataset.csv")

# Show first rows
print(df.head())

# Check basic info
print(df.info())

# Check missing values
print(df.isnull().sum())
```

```python
def categorize_addiction(score):
    if score <= 3:
        return "Low"
    elif score <= 6:
        return "Medium"
    else:
        return "High"

df["Addiction_Level"] = df["Addicted_Score"].apply(categorize_addiction)

# Check result
print(df[["Addicted_Score", "Addiction_Level"]].head())
```
